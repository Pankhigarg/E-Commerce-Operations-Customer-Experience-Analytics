```python
RAW_DATA_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "raw"
)

PROCESSED_DATA_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "processed"
)

REPORTS_PATH = os.path.join(
    PROJECT_ROOT,
    "reports"
)

# Create reports directory if it doesn't exist
os.makedirs(
    REPORTS_PATH,
    exist_ok=True
)

print("Raw data:")
print(RAW_DATA_PATH)

print("\nProcessed data:")
print(PROCESSED_DATA_PATH)

print("\nReports:")
print(REPORTS_PATH)
```

```python
raw_files = sorted(
    os.listdir(RAW_DATA_PATH)
)

print("Files in raw data folder:\n")

for file in raw_files:
    print("  ", file)
```

```python
tables = load_all_tables(
    RAW_DATA_PATH
)
```

```python
table_summary = get_table_summary(
    tables
)

display(
    table_summary
)
```

```python
AUDIT_FILE = os.path.join(
    REPORTS_PATH,
    "data_quality_audit.txt"
)

audit_database(
    tables,
    AUDIT_FILE
)
```

```python
for table_name, df in tables.items():

    print("\n" + "=" * 70)
    print(table_name)
    print("=" * 70)

    print(df.columns.tolist())
```

```python
for table_name, df in tables.items():

    print("\n" + "=" * 70)
    print(table_name)
    print("=" * 70)

    display(
        df.head(3)
    )
```

```python
null_summary = []

for table_name, df in tables.items():

    for column in df.columns:

        null_count = df[column].isnull().sum()

        null_pct = (
            df[column].isnull().mean() * 100
        )

        null_summary.append({
            "table": table_name,
            "column": column,
            "null_count": null_count,
            "null_percentage": round(null_pct, 2)
        })

null_summary = pd.DataFrame(
    null_summary
)

display(
    null_summary[
        null_summary["null_count"] > 0
    ].sort_values(
        "null_percentage",
        ascending=False
    )
)
```

```python
duplicate_summary = []

for table_name, df in tables.items():

    duplicate_summary.append({
        "table": table_name,
        "rows": len(df),
        "duplicate_rows": df.duplicated().sum(),
        "duplicate_percentage": round(
            df.duplicated().mean() * 100,
            2
        )
    })

duplicate_summary = pd.DataFrame(
    duplicate_summary
)

display(
    duplicate_summary
)
```

```python
summary_path = os.path.join(
    REPORTS_PATH,
    "table_summary.csv"
)

table_summary.to_csv(
    summary_path,
    index=False
)

print(
    f"✓ Table summary saved to: {summary_path}"
)
```
